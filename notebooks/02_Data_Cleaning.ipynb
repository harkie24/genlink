```python
import pandas as pd
```

```python
genes = pd.read_csv("/content/ensembl_genes.csv")
diseases = pd.read_csv("/content/rare_gene_disease_dataset.csv")
```

**Missing Value Analysis**

```python
genes_null_count = genes.isnull().sum()
diseases_null_count = diseases.isnull().sum()

print("Genes Dataset Null Count:\n", genes_null_count)
print("\nDiseases Dataset Null Count:\n", diseases_null_count)
```

**Statistics**


```python
print(f"Unique Diseases: {diseases['disease'].nunique()}")
```

**Disease Search Tool**

```python
disease_name = input("Enter disease name: ")

result = diseases[diseases['disease'].str.contains(disease_name, case = False)]
print(result)
```

**Gene Search Tool**

```python
gene_name = input("Enter gene name: ")

result = diseases[diseases['gene_symbol'].str.contains(gene_name, case=False)]

print(result)
```

**Top 20 Most Frequent Genes**

---

.value_counts() counts how many times each gene appears.

```python
top20_genes = (diseases['gene_symbol'].value_counts().head(20))
top20_genes
```

**Top 20 Diseases**

```python
top20_diseases = (diseases['disease'].value_counts().head(20))
top20_diseases
```
