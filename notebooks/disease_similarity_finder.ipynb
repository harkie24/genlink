```python
import pandas as pd
```

```python
genes = pd.read_csv("/workspaces/genlink/data/ensembl_genes.csv")
diseases = pd.read_csv("/workspaces/genlink/data/rare_gene_disease_dataset.csv")
```

using groupby(disease) pandas groups all rows having the same disease
using apply(set) python turns the genes into a set
Disease A -> {G1, G2, G3}
Disease B -> {G2, G3}
Disease C -> {G5}

```python
disease_genes = (
    diseases.groupby("disease")["gene_symbol"]
    .apply(set)
)
disease_genes
```

```python
disease_dict = disease_genes.to_dict()
```

& operator is used for intersection

```python
diseases_list = list(disease_dict.keys())
```

```python
similarities = []

for i in range(len(diseases)):
    for j in range(i + 1, len(diseases)):

        disease1 = diseases_list[i]
        disease2 = diseases_list[j]

        genes1 = disease_dict[disease1]
        genes2 = disease_dict[disease2]

        shared = genes1 & genes2

        similarities.append({
            "Disease 1": disease1,
            "Disease 2": disease2,
            "Shared Genes": len(shared),
            "Shared Gene List": ", ".join(sorted(shared))
        })
```

creating a dataframe

```python
similarity_df = pd.DataFrame(similarities)
```

sort

```python
similarity_df = similarity_df.sort_values(
    by="Shared Genes",
    ascending=False
)
```

removing unrelated diseases

```python
similarity_df = similarity_df[
    similarity_df["Shared Genes"] > 0
]
similarity_df
```
