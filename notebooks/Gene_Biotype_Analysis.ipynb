```python
import pandas as pd
```

```python
genes = pd.read_csv("/content/ensembl_genes.csv")
diseases = pd.read_csv("/content/rare_gene_disease_dataset.csv")
```

```python
merged = pd.merge(diseases, genes, left_on = 'gene_symbol', right_on = 'gene_symbol', how = 'inner')
merged.head()
```

**Top 5 Biotypes**

```python
top5_biotype = (merged['biotype'].value_counts())
top5_biotype
```

**Calculating percentages**

```python
biotype_df = (merged['biotype'].value_counts().reset_index())
biotype_df
```

```python
biotype_df['Percentage'] = (biotype_df['count']/biotype_df['count'].sum())*100
biotype_df
```

**Finding the Most Common Biotype**

idxmax() returns the index of the maximum

```python
most_common = (merged['biotype'].value_counts().idxmax())
print("Most common biotype:", most_common)
```

**Finding the Least Common Biotype**

```python
least_common = (merged['biotype'].value_counts().idxmin())
print("Least common biotype:", least_common)
```

**Filter only Protein-Coding Genes**

```python
protein_genes = merged[ merged['biotype'] == 'protein_coding']
protein_genes.head()
```

**Number of Diseases per Biotype**





```python
disease_per_biotype = (merged.groupby('biotype')['disease'].nunique().sort_values(ascending=False))
disease_per_biotype
```

**Complete Analysis**

.agg() allows multiple different calculations on different columns at the same time

```python
biotype_analysis = (merged.groupby('biotype').agg(
    Total_Genes = ('gene_symbol', 'nunique'),
    Total_Diseases = ('disease', 'nunique'),
).sort_values(
    by = 'Total_Genes',
    ascending = False
)
)
biotype_analysis
```
